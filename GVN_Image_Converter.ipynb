```python
import cv2
import numpy as np
import os
from google.colab import drive

# Mount Google Drive
drive.mount('/content/drive')
```

```python
def calculate_mean_gray_value(template):
    return np.mean(template)

def calculate_variance(pixel_grayVal, mean_gray_value):
    return np.mean((pixel_grayVal - mean_gray_value) ** 2)

def normalize_gray_scale_variance(gray_scale_variance_map):
    qmin = np.min(gray_scale_variance_map)
    qmax = np.max(gray_scale_variance_map)
    return (gray_scale_variance_map - qmin) / (qmax - qmin)

def gvn_processing(gray_image):
    rows, cols = gray_image.shape
    gray_scale_variance_map = np.zeros((rows, cols), dtype=np.float32)
    template_size = 3

    for i in range(1, rows - template_size):
        for j in range(1, cols - template_size):
            template = gray_image[i-1:i+(template_size-1), j-1:j+(template_size-1)]
            mean_gray_value = calculate_mean_gray_value(template)
            gray_scale_variance = calculate_variance(template, mean_gray_value)
            gray_scale_variance_map[i, j] = gray_scale_variance

    normalized_map = normalize_gray_scale_variance(gray_scale_variance_map)
    return normalized_map

```

```python
# Google Drive paths
base_path = "/content/drive/My Drive/Conductor_Image_Detection/Images/Lab_experiment/Classified"
original_base_path = os.path.join(base_path, "Original")
gvn_base_path = os.path.join(base_path, "GVN_Processed")

# Class folders
classes = ["Corrosion_Pollutant_Build_Up", "Fretting_Pollutant_Build_Up_Corrosion", "Healthy", "Minor_Abrasion", "Minor_Corrosion"]
#classes = ["Minor_Abrasion"]

subfolders = ["from_auto", "from_D", "from_H", "from_W"]
```

```python
for class_name in classes:
    class_original_path = os.path.join(original_base_path, class_name)
    class_gvn_path = os.path.join(gvn_base_path, class_name)
    os.makedirs(class_gvn_path, exist_ok=True)  # Ensure class folder exists

    print(f"\nProcessing class: {class_name}")

    for subfolder in subfolders:
        subfolder_original_path = os.path.join(class_original_path, subfolder)
        subfolder_gvn_path = os.path.join(class_gvn_path, subfolder)

        if os.path.exists(subfolder_original_path):
            os.makedirs(subfolder_gvn_path, exist_ok=True)  # Ensure subfolder exists in GVN_Processed

            images = os.listdir(subfolder_original_path)  # List images in subfolder

            if not images:
                print(f"  Skipping empty subfolder: {subfolder}")
                continue  # Skip empty subfolder

            print(f"  Processing subfolder: {subfolder} ({len(images)} images)")

            for filename in images:
                image_path = os.path.join(subfolder_original_path, filename)
                output_image_path = os.path.join(subfolder_gvn_path, filename)

                # Check if the processed file already exists
                if os.path.exists(output_image_path):
                    print(f"  Skipping {filename}: already processed.")
                    continue

                gray_image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)  # Convert to grayscale

                if gray_image is not None:
                    # Process the image
                    normalized_map = gvn_processing(gray_image)
                    normalized_map_uint8 = cv2.normalize(normalized_map, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

                    # Save the processed image in the correct subfolder
                    cv2.imwrite(output_image_path, normalized_map_uint8)

            print(f"  ✅ Finished processing subfolder: {subfolder}")

        else:
            print(f"  ❌ Subfolder '{subfolder}' does not exist, skipping.")

print("\n✅ Processing and saving completed!")
```
